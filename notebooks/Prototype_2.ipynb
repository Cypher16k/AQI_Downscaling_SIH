```python
# ============================================================
# PROTOTYPE 2 — DETERMINISTIC SPATIAL SPLIT
# NO RANDOM SPLITTING
# ============================================================

import numpy as np

# ------------------------------------------------------------
# Work at GRID level, not individual row level
# ------------------------------------------------------------

grid_locations = (
    data[["grid_id", "latitude", "longitude"]]
    .drop_duplicates("grid_id")
    .sort_values("longitude")
    .reset_index(drop=True)
)

print("=" * 80)
print("DETERMINISTIC SPATIAL SPLIT")
print("=" * 80)

print(f"Total grid cells : {len(grid_locations):,}")

# ------------------------------------------------------------
# Sort spatially from west → east
# ------------------------------------------------------------

n_grids = len(grid_locations)

train_end = int(n_grids * 0.70)
val_end = int(n_grids * 0.85)

train_grids = set(
    grid_locations.iloc[:train_end]["grid_id"]
)

val_grids = set(
    grid_locations.iloc[train_end:val_end]["grid_id"]
)

test_grids = set(
    grid_locations.iloc[val_end:]["grid_id"]
)

# ------------------------------------------------------------
# Assign rows using grid membership
# ------------------------------------------------------------

train_data = data[data["grid_id"].isin(train_grids)].copy()
val_data = data[data["grid_id"].isin(val_grids)].copy()
test_data = data[data["grid_id"].isin(test_grids)].copy()

# ------------------------------------------------------------
# Diagnostics
# ------------------------------------------------------------

print("\nGRID CELLS")
print(f"Train      : {len(train_grids):,}")
print(f"Validation : {len(val_grids):,}")
print(f"Test       : {len(test_grids):,}")

print("\nROWS")
print(f"Train      : {len(train_data):,}")
print(f"Validation : {len(val_data):,}")
print(f"Test       : {len(test_data):,}")

print("\nLONGITUDE RANGE")

print(
    f"Train      : "
    f"{train_data['longitude'].min():.6f} → "
    f"{train_data['longitude'].max():.6f}"
)

print(
    f"Validation : "
    f"{val_data['longitude'].min():.6f} → "
    f"{val_data['longitude'].max():.6f}"
)

print(
    f"Test       : "
    f"{test_data['longitude'].min():.6f} → "
    f"{test_data['longitude'].max():.6f}"
)

print("\nGRID OVERLAP CHECK")

print("Train ∩ Validation:",
      len(train_grids & val_grids))

print("Train ∩ Test:",
      len(train_grids & test_grids))

print("Validation ∩ Test:",
      len(val_grids & test_grids))

print("\nDATE COVERAGE")

print("Train:",
      train_data["date"].nunique())

print("Validation:",
      val_data["date"].nunique())

print("Test:",
      test_data["date"].nunique())

# Safety checks
assert len(train_grids & val_grids) == 0
assert len(train_grids & test_grids) == 0
assert len(val_grids & test_grids) == 0

print("\n✓ NO RANDOM SPLIT")
print("✓ NO GRID-ID OVERLAP")
print("✓ ALL DATES OF EACH GRID STAY TOGETHER")
```

```python
# ============================================================
# PROTOTYPE 2 — XGBOOST TUNING
# Validation-driven tuning
# TEST SET REMAINS UNTOUCHED
# ============================================================

import numpy as np
import pandas as pd
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from itertools import product
from time import time

# ------------------------------------------------------------
# 1. TUNING GRID
# ------------------------------------------------------------

param_grid = {
    "n_estimators": [300, 600],
    "max_depth": [3, 5, 7],
    "learning_rate": [0.03, 0.05],
    "min_child_weight": [1, 5],
    "subsample": [0.8],
    "colsample_bytree": [0.8],
    "gamma": [0],
    "reg_alpha": [0],
    "reg_lambda": [1, 5]
}

keys = list(param_grid.keys())
combinations = list(product(*[param_grid[k] for k in keys]))

print("=" * 80)
print("XGBOOST HYPERPARAMETER TUNING")
print("=" * 80)
print(f"Configurations to test : {len(combinations)}")
print("Selection metric       : Validation RMSE")
print("Test set               : LOCKED / UNTOUCHED")


# ------------------------------------------------------------
# 2. RUN VALIDATION SEARCH
# ------------------------------------------------------------

tuning_results = []

for i, values in enumerate(combinations, 1):

    params = dict(zip(keys, values))

    start = time()

    model = XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1,
        **params
    )

    model.fit(
        X_train,
        y_train,
        verbose=False
    )

    pred_val = model.predict(X_val)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred_val)
    )

    mae = mean_absolute_error(
        y_val, pred_val
    )

    r2 = r2_score(
        y_val, pred_val
    )

    tuning_results.append({
        **params,
        "Validation_RMSE": rmse,
        "Validation_MAE": mae,
        "Validation_R2": r2,
        "Time_sec": time() - start
    })

    print(
        f"[{i:03d}/{len(combinations)}] "
        f"RMSE={rmse:.10f} | "
        f"R²={r2:.6f}"
    )


# ------------------------------------------------------------
# 3. SORT RESULTS
# ------------------------------------------------------------

xgb_tuning_results = (
    pd.DataFrame(tuning_results)
    .sort_values("Validation_RMSE")
    .reset_index(drop=True)
)

print("\n" + "=" * 80)
print("XGBOOST TUNING RESULTS — TOP 10")
print("=" * 80)

display(
    xgb_tuning_results.head(10)
)
```

```python
# ============================================================
# PROTOTYPE 2 — XGBOOST FINE TUNING
# Focused search around the best validation configuration
# TEST SET REMAINS COMPLETELY UNTOUCHED
# ============================================================

import numpy as np
import pandas as pd
from itertools import product
from time import time

from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


# ------------------------------------------------------------
# 1. FOCUSED SEARCH SPACE
# ------------------------------------------------------------

fine_grid = {
    "n_estimators": [600, 900, 1200],
    "max_depth": [6, 7, 8],
    "learning_rate": [0.02, 0.03, 0.05],
    "min_child_weight": [1, 3, 5, 8],
    "subsample": [0.75, 0.85],
    "colsample_bytree": [0.75, 0.85],
    "reg_lambda": [1, 3, 5]
}

keys = list(fine_grid.keys())

fine_combinations = list(
    product(*[fine_grid[k] for k in keys])
)

print("=" * 80)
print("XGBOOST FINE-TUNING")
print("=" * 80)

print(f"Configurations : {len(fine_combinations)}")
print("Metric         : Validation RMSE")
print("Test set       : LOCKED")


# ------------------------------------------------------------
# 2. RUN FINE-TUNING
# ------------------------------------------------------------

fine_results = []

for i, values in enumerate(fine_combinations, 1):

    params = dict(zip(keys, values))

    start = time()

    model = XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1,
        gamma=0,
        reg_alpha=0,
        **params
    )

    model.fit(
        X_train,
        y_train,
        verbose=False
    )

    pred_val = model.predict(X_val)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred_val)
    )

    mae = mean_absolute_error(
        y_val, pred_val
    )

    r2 = r2_score(
        y_val, pred_val
    )

    fine_results.append({
        **params,
        "Validation_RMSE": rmse,
        "Validation_MAE": mae,
        "Validation_R2": r2,
        "Time_sec": time() - start
    })

    print(
        f"[{i:03d}/{len(fine_combinations)}] "
        f"RMSE={rmse:.10f} | "
        f"R²={r2:.6f}"
    )


# ------------------------------------------------------------
# 3. RANK RESULTS
# ------------------------------------------------------------

xgb_fine_results = (
    pd.DataFrame(fine_results)
    .sort_values(
        ["Validation_RMSE", "Validation_MAE"],
        ascending=True
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. TOP 10
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("XGBOOST FINE-TUNING — TOP 10")
print("=" * 80)

display(
    xgb_fine_results.head(10)
)
```

```python
# ============================================================
# PROTOTYPE 2 — CYCLIC SPATIAL FEATURES
# ============================================================

import numpy as np

def add_spatial_features(df):
    df = df.copy()

    # Convert degrees → radians
    lat_rad = np.radians(df["latitude"])
    lon_rad = np.radians(df["longitude"])

    # Cyclic coordinate representation
    df["latitude_sin"] = np.sin(lat_rad)
    df["latitude_cos"] = np.cos(lat_rad)

    df["longitude_sin"] = np.sin(lon_rad)
    df["longitude_cos"] = np.cos(lon_rad)

    return df


train_spatial = add_spatial_features(train_data)
val_spatial   = add_spatial_features(val_data)
test_spatial  = add_spatial_features(test_data)


print("=" * 80)
print("CYCLIC SPATIAL FEATURES")
print("=" * 80)

print("Added:")
print("  latitude_sin")
print("  latitude_cos")
print("  longitude_sin")
print("  longitude_cos")

print("\nTrain shape :", train_spatial.shape)
print("Validation  :", val_spatial.shape)
print("Test        :", test_spatial.shape)
```

```python
# ============================================================
# PROTOTYPE 2 — 4-WAY SPATIAL LEAVE-ONE-QUADRANT-OUT
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

print("=" * 80)
print("PROTOTYPE 2 — 4-WAY SPATIAL VALIDATION")
print("=" * 80)

# ------------------------------------------------------------
# QUADRANT DEFINITIONS
# ------------------------------------------------------------

quadrants = ["NW", "NE", "SW", "SE"]

experiments = {
    "SE": ["NW", "NE", "SW"],
    "SW": ["NW", "NE", "SE"],
    "NE": ["NW", "SW", "SE"],
    "NW": ["NE", "SW", "SE"],
}

results = []

# ------------------------------------------------------------
# GRID REFERENCE
# ------------------------------------------------------------

grid_ref = (
    df[["grid_id", "latitude", "longitude"]]
    .drop_duplicates("grid_id")
    .copy()
)

# Use the SAME midpoint used for the diagonal experiments
lat_mid = df["latitude"].median()
lon_mid = df["longitude"].median()

grid_ref["quadrant"] = np.select(
    [
        (grid_ref["latitude"] >= lat_mid) &
        (grid_ref["longitude"] < lon_mid),

        (grid_ref["latitude"] >= lat_mid) &
        (grid_ref["longitude"] >= lon_mid),

        (grid_ref["latitude"] < lat_mid) &
        (grid_ref["longitude"] < lon_mid),

        (grid_ref["latitude"] < lat_mid) &
        (grid_ref["longitude"] >= lon_mid),
    ],
    ["NW", "NE", "SW", "SE"],
    default="UNKNOWN"
)

# ------------------------------------------------------------
# RUN FOUR EXPERIMENTS
# ------------------------------------------------------------

for test_quadrant, train_quadrants in experiments.items():

    print("\n" + "=" * 80)
    print(f"TEST QUADRANT: {test_quadrant}")
    print(f"TRAIN QUADRANTS: {' + '.join(train_quadrants)}")
    print("=" * 80)

    # --------------------------------------------
    # GRID SPLIT
    # --------------------------------------------

    train_grids = set(
        grid_ref.loc[
            grid_ref["quadrant"].isin(train_quadrants),
            "grid_id"
        ]
    )

    test_grids = set(
        grid_ref.loc[
            grid_ref["quadrant"] == test_quadrant,
            "grid_id"
        ]
    )

    overlap = train_grids.intersection(test_grids)

    assert len(overlap) == 0, (
        f"GRID LEAKAGE! {len(overlap)} overlapping grids"
    )

    train_split = df[df["grid_id"].isin(train_grids)].copy()
    test_split = df[df["grid_id"].isin(test_grids)].copy()

    # --------------------------------------------
    # TARGET
    # --------------------------------------------

    y_train = train_split["satellite_no2"]
    y_test = test_split["satellite_no2"]

    # --------------------------------------------
    # FEATURES
    # --------------------------------------------

    X_train = train_split[feature_cols]
    X_test = test_split[feature_cols]

    assert X_train.isna().sum().sum() == 0
    assert X_test.isna().sum().sum() == 0
    assert y_train.isna().sum() == 0
    assert y_test.isna().sum() == 0

    # --------------------------------------------
    # FRESH MODEL
    # --------------------------------------------

    model = ExtraTreesRegressor(
        n_estimators=500,
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features=1.0,
        random_state=42,
        n_jobs=-1
    )

    print(f"Train samples : {len(train_split):,}")
    print(f"Test samples  : {len(test_split):,}")
    print(f"Train grids   : {len(train_grids):,}")
    print(f"Test grids    : {len(test_grids):,}")

    print("\nTraining...")
    model.fit(X_train, y_train)

    # --------------------------------------------
    # PREDICTION
    # --------------------------------------------

    y_pred = model.predict(X_test)

    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )

    mae = mean_absolute_error(
        y_test, y_pred
    )

    r2 = r2_score(
        y_test, y_pred
    )

    # --------------------------------------------
    # BASELINE
    # --------------------------------------------

    baseline_pred = np.full(
        len(y_test),
        y_train.mean()
    )

    baseline_rmse = np.sqrt(
        mean_squared_error(
            y_test,
            baseline_pred
        )
    )

    baseline_mae = mean_absolute_error(
        y_test,
        baseline_pred
    )

    baseline_r2 = r2_score(
        y_test,
        baseline_pred
    )

    rmse_improvement = (
        (baseline_rmse - rmse)
        / baseline_rmse * 100
    )

    mae_improvement = (
        (baseline_mae - mae)
        / baseline_mae * 100
    )

    # --------------------------------------------
    # OUTPUT
    # --------------------------------------------

    print("\nRESULT")
    print("-" * 60)

    print(f"RMSE : {rmse:.10f}")
    print(f"MAE  : {mae:.10f}")
    print(f"R²   : {r2:.6f}")

    print("\nBaseline")
    print(f"RMSE : {baseline_rmse:.10f}")
    print(f"MAE  : {baseline_mae:.10f}")
    print(f"R²   : {baseline_r2:.6f}")

    print("\nImprovement")
    print(f"RMSE : {rmse_improvement:.2f}%")
    print(f"MAE  : {mae_improvement:.2f}%")

    # --------------------------------------------
    # STORE
    # --------------------------------------------

    results.append({
        "test_quadrant": test_quadrant,
        "train_quadrants": "+".join(train_quadrants),
        "train_samples": len(train_split),
        "test_samples": len(test_split),
        "train_grids": len(train_grids),
        "test_grids": len(test_grids),
        "RMSE": rmse,
        "MAE": mae,
        "R2": r2,
        "baseline_RMSE": baseline_rmse,
        "baseline_MAE": baseline_mae,
        "baseline_R2": baseline_r2,
        "RMSE_improvement_%": rmse_improvement,
        "MAE_improvement_%": mae_improvement
    })


# ============================================================
# FINAL LEADERBOARD
# ============================================================

results_df = pd.DataFrame(results)

print("\n\n" + "=" * 80)
print("4-WAY SPATIAL VALIDATION — FINAL LEADERBOARD")
print("=" * 80)

display(
    results_df[
        [
            "test_quadrant",
            "train_quadrants",
            "test_samples",
            "test_grids",
            "RMSE",
            "MAE",
            "R2",
            "RMSE_improvement_%",
            "MAE_improvement_%"
        ]
    ].sort_values(
        "R2",
        ascending=False
    )
)


# ============================================================
# OVERALL SPATIAL GENERALIZATION
# ============================================================

print("\n" + "=" * 80)
print("OVERALL SPATIAL GENERALIZATION")
print("=" * 80)

print(
    f"Mean R² : {results_df['R2'].mean():.6f}"
)

print(
    f"Std R²  : {results_df['R2'].std():.6f}"
)

print(
    f"Min R²  : {results_df['R2'].min():.6f}"
)

print(
    f"Max R²  : {results_df['R2'].max():.6f}"
)

print(
    f"Mean RMSE improvement : "
    f"{results_df['RMSE_improvement_%'].mean():.2f}%"
)

print(
    f"Mean MAE improvement  : "
    f"{results_df['MAE_improvement_%'].mean():.2f}%"
)

# ------------------------------------------------------------
# LEAKAGE GUARANTEE
# ------------------------------------------------------------

assert all(
    results_df["test_grids"] > 0
)

print("\n✓ FOUR SPATIAL TESTS COMPLETED")
print("✓ EACH TEST QUADRANT WAS HELD OUT")
print("✓ NO GRID OVERLAP")
print("✓ FRESH MODEL FOR EACH EXPERIMENT")
```

```python
# ============================================================
# PROTOTYPE 2 — COMPLETE SPATIAL PREDICTION EXPORT
# ============================================================
#
# Creates ALL six prediction CSVs:
#
#   1. diagonal
#   2. reverse diagonal
#   3. 4-way NE
#   4. 4-way SW
#   5. 4-way SE
#   6. 4-way NW
#
# IMPORTANT:
# - Fresh Extra Trees model for EVERY experiment
# - Zero grid overlap
# - No random row split
# - Target = satellite_no2
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

OUT = "/content/prototype2_outputs_complete"
os.makedirs(OUT, exist_ok=True)

# ------------------------------------------------------------
# MASTER DATA
# ------------------------------------------------------------

# Reload data to ensure it is available in this execution context
DATA_PATH = "/content/Prototype2_ML_Master_91030.csv"
data = pd.read_csv(DATA_PATH)

master = data.copy()

master["date"] = pd.to_datetime(master["date"])

TARGET = "satellite_no2"

FEATURES = [
    "aod",
    "cloud",
    "boundary_layer_height",
    "dew_point",
    "precipitation",
    "relative_humidity",
    "solar_radiation",
    "surface_pressure",
    "temperature",
    "wind_direction",
    "wind_speed",
    "bare_fraction",
    "builtup_fraction",
    "cropland_fraction",
    "vegetation_fraction",
    "water_fraction",
    "ndvi",
    "elevation_m",
    "distance_to_water_m",
    "built_up_density",
    "nighttime_lights",
    "population_density",
    "road_density_km_per_km2",
    "distance_to_road_m"
]

# ------------------------------------------------------------
# VERIFY
# ------------------------------------------------------------

assert len(master) == 91030
assert master[FEATURES].isna().sum().sum() == 0
assert master[TARGET].isna().sum() == 0

# ------------------------------------------------------------
# SPATIAL QUADRANTS
# Same midpoint methodology used in Prototype 2
# ------------------------------------------------------------

lat_mid = master["latitude"].median()
lon_mid = master["longitude"].median()

def assign_quadrant(row):

    if row["latitude"] >= lat_mid and row["longitude"] >= lon_mid:
        return "NE"

    elif row["latitude"] >= lat_mid and row["longitude"] < lon_mid:
        return "NW"

    elif row["latitude"] < lat_mid and row["longitude"] >= lon_mid:
        return "SE"

    else:
        return "SW"


master["quadrant"] = master.apply(assign_quadrant, axis=1)

print("=" * 80)
print("PROTOTYPE 2 — COMPLETE PREDICTION EXPORT")
print("=" * 80)

print("\nQUADRANT COUNTS:")
print(master["quadrant"].value_counts())


# ============================================================
# MODEL
# ============================================================
#
# IMPORTANT:
# This is a FRESH MODEL for every experiment.
#
# If your final Prototype 2 Extra Trees used different
# hyperparameters, replace ONLY this function with those
# exact parameters.
# ============================================================

def new_extra_trees():

    return ExtraTreesRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1,
        max_features=1.0,
        min_samples_leaf=1
    )


# ============================================================
# EXPORT FUNCTION
# ============================================================

def run_experiment(train_df, test_df, experiment_name):

    print("\n" + "=" * 80)
    print(f"EXPERIMENT: {experiment_name}")
    print("=" * 80)

    # --------------------------------------------------------
    # Spatial integrity
    # --------------------------------------------------------

    train_grids = set(train_df["grid_id"].unique())
    test_grids = set(test_df["grid_id"].unique())

    overlap = train_grids.intersection(test_grids)

    print(f"Train rows : {len(train_df):,}")
    print(f"Test rows  : {len(test_df):,}")
    print(f"Train grids: {len(train_grids):,}")
    print(f"Test grids : {len(test_grids):,}")
    print(f"Grid overlap: {len(overlap)}")

    assert len(overlap) == 0

    # --------------------------------------------------------
    # Model
    # --------------------------------------------------------

    X_train = train_df[FEATURES]
    y_train_local = train_df[TARGET]

    X_test = test_df[FEATURES]
    y_test_local = test_df[TARGET]

    model = new_extra_trees()

    print("\nTraining fresh Extra Trees...")

    model.fit(X_train, y_train_local)

    print("✓ Training complete")

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    predictions = model.predict(X_test)

    rmse = np.sqrt(
        mean_squared_error(y_test_local, predictions)
    )

    mae = mean_absolute_error(
        y_test_local,
        predictions
    )

    r2 = r2_score(
        y_test_local,
        predictions
    )

    print("\nRESULT")
    print("-" * 60)
    print(f"RMSE : {rmse:.12f}")
    print(f"MAE  : {mae:.12f}")
    print(f"R²   : {r2:.6f}")

    # --------------------------------------------------------
    # GitHub-ready prediction table
    # --------------------------------------------------------

    result = pd.DataFrame({

        "target_id":
            test_df["grid_id"].values,

        "date":
            test_df["date"].values,

        "latitude":
            test_df["latitude"].values,

        "longitude":
            test_df["longitude"].values,

        "actual_no2":
            y_test_local.values,

        "predicted_no2":
            predictions,

        "ndvi":
            test_df["ndvi"].values,

        "population_density":
            test_df["population_density"].values,

        "aod":
            test_df["aod"].values,

        "cloud":
            test_df["cloud"].values,

        "experiment":
            experiment_name,

        "r2":
            r2,

        "rmse":
            rmse,

        "mae":
            mae
    })

    # Time fields
    result["year"] = result["date"].dt.year
    result["month"] = result["date"].dt.month
    result["day_of_year"] = result["date"].dt.dayofyear

    # Error diagnostics
    result["error_no2"] = (
        result["predicted_no2"] -
        result["actual_no2"]
    )

    result["absolute_error_no2"] = (
        result["error_no2"].abs()
    )

    # --------------------------------------------------------
    # Integrity
    # --------------------------------------------------------

    assert len(result) == len(test_df)
    assert result["actual_no2"].isna().sum() == 0
    assert result["predicted_no2"].isna().sum() == 0

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    filename = (
        experiment_name.lower()
        .replace(" ", "_")
        .replace("+", "plus")
        + "_predictions.csv"
    )

    path = os.path.join(OUT, filename)

    result.to_csv(path, index=False)

    print(f"\n✓ SAVED:")
    print(path)

    print(
        f"File size: "
        f"{os.path.getsize(path)/(1024**2):.2f} MB"
    )

    return result, model


# ============================================================
# 1 — DIAGONAL
# ============================================================
#
# TRAIN = NE + SW
# TEST  = NW + SE
#
# Expected:
# train ≈ 41,423
# test  ≈ 49,607
# ============================================================

diagonal_train_final = master[
    master["quadrant"].isin(["NE", "SW"])
].copy()

diagonal_test_final = master[
    master["quadrant"].isin(["NW", "SE"])
].copy()

diagonal_result, diagonal_model = run_experiment(
    diagonal_train_final,
    diagonal_test_final,
    "diagonal_NE_SW_to_NW_SE"
)


# ============================================================
# 2 — REVERSE DIAGONAL
# ============================================================
#
# TRAIN = NW + SE
# TEST  = NE + SW
#
# Expected:
# train ≈ 49,607
# test  ≈ 41,423
# ============================================================

reverse_train_final = master[
    master["quadrant"].isin(["NW", "SE"])
].copy()

reverse_test_final = master[
    master["quadrant"].isin(["NE", "SW"])
].copy()

reverse_result, reverse_model = run_experiment(
    reverse_train_final,
    reverse_test_final,
    "reverse_diagonal_NW_SE_to_NE_SW"
)


# ============================================================
# 3–6 — FOUR-WAY SPATIAL VALIDATION
# ============================================================

quadrants = ["NE", "SW", "SE", "NW"]

four_way_results = []

for test_quad in quadrants:

    train_quads = [
        q for q in quadrants
        if q != test_quad
    ]

    train_df = master[
        master["quadrant"].isin(train_quads)
    ].copy()

    test_df = master[
        master["quadrant"] == test_quad
    ].copy()

    result, model = run_experiment(
        train_df,
        test_df,
        f"4way_test_{test_quad}"
    )

    four_way_results.append({
        "test_quadrant": test_quad,
        "train_quadrants": "+".join(train_quads),
        "test_samples": len(test_df),
        "test_grids": test_df["grid_id"].nunique(),
        "RMSE": np.sqrt(
            mean_squared_error(
                test_df[TARGET],
                result["predicted_no2"]
            )
        ),
        "MAE": mean_absolute_error(
            test_df[TARGET],
            result["predicted_no2"]
        ),
        "R2": r2_score(
            test_df[TARGET],
            result["predicted_no2"]
        )
    })


# ============================================================
# LEADERBOARD
# ============================================================

leaderboard = pd.DataFrame(four_way_results)

leaderboard = leaderboard.sort_values(
    "R2",
    ascending=False
).reset_index(drop=True)

leaderboard_path = os.path.join(
    OUT,
    "prototype2_4way_spatial_leaderboard.csv"
)

leaderboard.to_csv(
    leaderboard_path,
    index=False
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("PROTOTYPE 2 — COMPLETE EXPORT FINISHED")
print("=" * 80)

print("\n4-WAY LEADERBOARD:")
print(leaderboard.to_string(index=False))

print("\n" + "-" * 80)
print("FILES")
print("-" * 80)

for filename in sorted(os.listdir(OUT)):

    path = os.path.join(OUT, filename)

    if os.path.isfile(path):

        print(
            f"{filename:<55} "
            f"{os.path.getsize(path)/(1024**2):>8.2f} MB"
        )

print("\n✓ DIAGONAL EXPORTED")
print("✓ REVERSE DIAGONAL EXPORTED")
print("✓ NE EXPORTED")
print("✓ SW EXPORTED")
print("✓ SE EXPORTED")
print("✓ NW EXPORTED")
print("✓ 4-WAY LEADERBOARD EXPORTED")

print("\nPrototype 2 prediction archive complete.")
```
